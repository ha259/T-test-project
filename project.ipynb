```python
import pandas as pd  
import matplotlib.pyplot as plt 
import seaborn as sns 
import scipy.stats as stats 
df=pd.read_csv("dataset2.csv",encoding="latin-1")
df.head()
```

```python
df=df.drop(columns=df.columns[df.isna().mean()>0.9])
df=df.drop("Patient ID",axis=1)
df.shape
```

```python
print(df["SARS-Cov-2 exam result"].value_counts(normalize=True)*100)
```

```python
floatcolonnes=df.select_dtypes(include="float")
objectcolonnes=df.select_dtypes(include="object")
for col in floatcolonnes.keys():
    sns.distplot(df[col])
    plt.show()
```

```python
for col in floatcolonnes.keys():

    sns.distplot(df[df["SARS-Cov-2 exam result"]=="positive"][col],label="positive",color="blue",kde=True)
    sns.distplot(df[df["SARS-Cov-2 exam result"]=="negative"][col],label="negative",color="red",kde=True)
    plt.legend()
    plt.show()
```

```python
for col in objectcolonnes.keys():
    df["est_malade"]=(df["SARS-Cov-2 exam result"]=="positive") & (df[col]=="detected")
```

```python
sns.heatmap(floatcolonnes.corr())
```

T test

```python
print("columns that significantly different:")
for col in floatcolonnes:
    ttest=stats.ttest_ind(df[df["SARS-Cov-2 exam result"]=="positive"][col],df[df["SARS-Cov-2 exam result"]=="negative"][col],nan_policy="omit")
    if ttest.pvalue <0.05:
        print(col)
```
